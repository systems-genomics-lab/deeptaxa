# DeepTaxa Tutorial: Classifying Full-Length 16S ASVs to the Species Level and Validating Against a Mock Community

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/systems-genomics-lab/deeptaxa/blob/main/tutorials/validation/full-length.ipynb)

Many microbiome questions need species-level resolution, not just genus, but a single
hypervariable region rarely carries enough signal to reach it. The full-length 16S gene
carries more, so full-length sequencing, for example with PacBio circular consensus
reads, allows finer assignment. This notebook examines that with DeepTaxa on
full-length ASVs from the ZymoBIOMICS standard, assessing how reliably species can
be named when the composition is already known. The steps are the same as for V3-V4, with
the full-length checkpoint and a closer look at species.

The data is the ZymoBIOMICS standard sequenced as full-length 16S on a PacBio Revio
instrument (accession SRR38028391). The analysis begins from ASVs produced with the
DADA2 PacBio workflow, summarized in the Appendix. The eight species are known, so
accuracy can be measured directly.

## The sequences

These ASVs come from the same ZymoBIOMICS standard sequenced over the complete 16S
gene, from the public SRA run SRR38028391. A single genome carries several slightly
different copies of the gene, which full-length sequencing resolves, so there are more
ASVs than species. The Appendix shows how they were produced.

````python
import urllib.request
base = "https://raw.githubusercontent.com/systems-genomics-lab/deeptaxa/main/tutorials/validation/data"
urllib.request.urlretrieve(f"{base}/fulllength_seqs.fasta", "asvs.fasta")
counts = pd.read_csv(f"{base}/fulllength_counts.tsv", sep="\t", index_col=0)["count"]
print(len(counts), "full-length ASVs")
````

## Download the full-length checkpoint

The reads span the full-length gene, so the full-length checkpoint is used. Applying
the V3-V4 checkpoint here would be an off-region use and would lower accuracy.

````python
!wget -q -nc https://huggingface.co/systems-genomics-lab/deeptaxa/resolve/main/deeptaxa-full-length-v1.pt
````

## Classify

````python
!deeptaxa predict --fasta-file asvs.fasta --checkpoint deeptaxa-full-length-v1.pt --output-dir out --tabular
pred = pd.read_csv("out/asvs_deeptaxa_predictions.tsv", sep="\t").set_index("sequence_id")
print("mean confidence: genus", round(pred["genus_raw_score"].mean(), 2),
      "species", round(pred["species_raw_score"].mean(), 2))
````

On full-length input the confidence is high at every rank, including species. This
is the opposite of what the same model produces on short fragments, and it shows that
the score reflects how much information the input carries.

## Validate against the known genera

As in the V3-V4 tutorial, GTDB names are reduced to plain genus names so they can be
compared with the NCBI names of the standard, and a reference result is loaded for
cross-checking: DADA2 with the SILVA v138 database, labeled DADA2 + SILVA below. SILVA
is the reference database; DADA2 is the classifier, using a naive Bayes model for the
genus and exact matching for the species.

````python
import re
silva = pd.read_csv(f"{base}/fulllength_silva.tsv", sep="\t").set_index("ASV").reindex(pred.index)

def base_genus(name):
    if not isinstance(name, str):                      # missing genus (NaN)
        return None
    name = re.sub(r"[ _-].*", "", name).lower()        # drop GTDB suffixes and "sensu stricto"
    return {"limosilactobacillus": "lactobacillus", "shigella": "escherichia"}.get(name, name)

EXPECTED = {"pseudomonas", "escherichia", "salmonella", "lactobacillus",
            "enterococcus", "staphylococcus", "listeria", "bacillus"}

dt_genus = pred["genus_predicted"].map(base_genus)
sv_genus = silva["Genus"].map(base_genus)

print("DeepTaxa recovered", len(set(dt_genus) & EXPECTED), "of 8 genera")
print("DADA2 + SILVA recovered", len(set(sv_genus.dropna()) & EXPECTED), "of 8 genera")
print("They agree on the genus of",
      round(counts[dt_genus == sv_genus].sum() / counts.sum() * 100), "% of reads")
````

## Species: the main advantage of full-length data

The table lists the species calls for the most abundant ASVs; a single-word entry is a
genus only, where the method stopped, and a two-word entry is a species. From the
full-length gene DeepTaxa assigns a species to most ASVs, while DADA2 + SILVA assigns a
species to a much smaller share and otherwise stops at the genus. The reference step is
exact matching, which assigns a species only on an exact, unambiguous reference match,
so its genus-only entries are reads it declined to name rather than reads it named
incorrectly, and every species it does assign is correct. DeepTaxa still stops at the
genus on a few ASVs, such as the most abundant *Staphylococcus*, but it resolves a
species far more often than the reference. Where it names a species, the call is almost
always a close relative of the expected organism, for example a member of the *Bacillus
subtilis* group or *Shigella* for *Escherichia coli*, rather than an unrelated error.
The confidence scores back this up: full-length input lets DeepTaxa assign a species
to far more reads than a short region allows, or than the exact-matching reference
resolves. The reference is exactly correct
whenever it commits, and on a few ASVs, such as the *Salmonella* here, it resolves the
species that DeepTaxa misses, so the gain is in coverage rather than uniform
superiority.

````python
pd.DataFrame({
    "reads": counts,
    "DeepTaxa": (pred["species_predicted"]
                 .str.replace("Unclassified_", "", regex=False)
                 .str.replace(r"_[A-Za-z0-9]+", "", regex=True)),
    "DADA2 + SILVA": (silva["Genus"].fillna("") + " " + silva["Species"].fillna("")).str.strip(),
}).sort_values("reads", ascending=False).head(10)
````

## Summary

On full-length sequences DeepTaxa recovered all eight genera, agreed with the
reference method at the genus rank, and assigned a species to most ASVs with high
confidence, stopping at the genus only where the marker does not separate the
organisms. Its non-exact species calls are close relatives that 16S cannot
resolve, so species labels remain subject to the limited species resolution of the
marker, but the coverage is far greater than an exact-matching reference method
provides. The practical recommendation is to use the full-length checkpoint for
full-length data.

## Appendix: how the ASVs were made

The ASVs were produced with the DADA2 workflow for PacBio circular consensus reads.

```r
library(dada2)
nop <- removePrimers("ccs.fastq.gz", "noprimer.fastq.gz",
                     primer.fwd = "AGRGTTYGATYMTGGCTCAG",
                     primer.rev = dada2:::rc("RGYTACCTTGTTACGACTT"), orient = TRUE)
filterAndTrim("noprimer.fastq.gz", "filt.fastq.gz",
              minLen = 1000, maxLen = 1600, maxN = 0, maxEE = 2, multithread = TRUE)
drp <- derepFastq("filt.fastq.gz")
err <- learnErrors(drp, errorEstimationFunction = dada2:::PacBioErrfun, BAND_SIZE = 32, multithread = TRUE)
dd  <- dada(drp, err = err, BAND_SIZE = 32, multithread = TRUE)
st  <- removeBimeraDenovo(makeSequenceTable(dd), method = "consensus", multithread = TRUE)
```
